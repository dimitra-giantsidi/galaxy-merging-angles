```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
golden_ratio = (1 + np.sqrt(5))/2
print(golden_ratio)
```

```python
n = 48
```

```python
i_array = np.linspace(0,n-1,n)
print(i_array)
```

```python
z_array = 1 - 2*i_array/(n-1)
print(z_array)
```

```python
radius_array = np.sqrt(1-z_array**2)
print(radius_array)
```

```python
longitude_array = 2*np.pi * i_array/golden_ratio
latitude_array = np.arcsin(z_array)
print(longitude_array)
print(latitude_array)


```

```python
fig, ax = plt.subplots()
ax.scatter(np.mod(longitude_array,2*np.pi), latitude_array)
```

```python
x_array = [0] * n
y_array = [0] * n

for k in range(n):
    x_array[k] = radius_array[k] * np.cos(longitude_array[k])
    y_array[k] = radius_array[k] * np.sin(longitude_array[k])

print(x_array)
print(radius_array)
print(np.cos(longitude_array))
```

```python
fig, ax = plt.subplots()

ax.scatter(x_array, y_array)
circle = plt.Circle((0, 0), radius=1, color="red", fill=False, linewidth=2, zorder=5)
ax.add_patch(circle)

plt.show()
```

```python
fig, ax = plt.subplots()

ax.scatter(y_array, z_array)
circle = plt.Circle((0, 0), radius=1, color="red", fill=False, linewidth=2, zorder=5)
ax.add_patch(circle)

plt.show()
```

```python
radius=1

u = np.linspace(0, 2 * np.pi, 30)
v = np.linspace(0, np.pi, 30)
x_sphere = radius * np.outer(np.cos(u), np.sin(v))
y_sphere = radius * np.outer(np.sin(u), np.sin(v))
z_sphere = radius * np.outer(np.ones_like(u), np.cos(v))

x_arr = np.asarray(x_array)
y_arr = np.asarray(y_array)
z_arr = np.asarray(z_array)

fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(projection='3d')

elev, azim = 30, 45
ax.view_init(elev=elev, azim=azim)

elev_rad = np.radians(elev)
azim_rad = np.radians(azim)

cam_x = np.cos(elev_rad) * np.cos(azim_rad)
cam_y = np.cos(elev_rad) * np.sin(azim_rad)
cam_z = np.sin(elev_rad)
camera_vector = np.array([cam_x, cam_y, cam_z])

points_matrix = np.vstack((x_arr, y_arr, z_arr)).T
dot_products = np.dot(points_matrix, camera_vector)
visible_mask = dot_products > 0

x_visible = x_arr[visible_mask]
y_visible = y_arr[visible_mask]
z_visible = z_arr[visible_mask]

ax.plot_wireframe(x_sphere, y_sphere, z_sphere, color='green', alpha=0.3, linewidth=0.5)
scatter = ax.scatter3D(x_visible, y_visible, z_visible, color='blue', s=40, edgecolors='black')

ax.set_box_aspect([1,1,1])
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
plt.title('Points on Fibonacci Unit Sphere: n=48')

plt.show()
```
